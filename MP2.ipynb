```python
from pathlib import Path  
from requests import Session

base_url = "https://datasets.imdbws.com/"
data_dir = Path.home() / "Downloads"
def download_imdb(dest_file):
    url = base_url + dest_file 
    local_path = data_dir / dest_file
    if local_path.exists():
        print(f"{local_path} existe déjà. Téléchargement ignoré.")
        return 
    s = Session()
    with s.get(url, stream=True) as r:
        r.raise_for_status()
        with open(local_path, "wb") as f:
            for chunk in r.iter_content(chunk_size=8192):
                if chunk: 
                    f.write(chunk)
```

```python
def get_actor(i):
    # 1. On récupère les données brutes de l'acteur à la position i
    # Structure de actors[i] : [Nom, [Liste des index films]]
    actor_data = actors[i]
    
    name = actor_data[0]
    movie_indices = actor_data[1] # C'est une liste d'entiers ex: [543, 12, 0]
    
    # 2. On prépare la liste détaillée des films
    movies_list = []
    
    # Pour chaque numéro de film dans lequel l'acteur a joué
    for index_movie in movie_indices:
        # On va chercher les infos du film dans la grande liste 'movies'
        # Structure de movies[x] : [Titre, [Liste Acteurs], Année]
        movie_entry = movies[index_movie]
        
        movie_title = movie_entry[0]
        movie_year = movie_entry[2]
        
        # On crée le petit dictionnaire pour ce film
        movie_dict = {
            'index': i,
            'name': movie_title,
            'year': movie_year,
            'index': index_movie
        }
        movies_list.append(movie_dict)
    
    # 3. On retourne le dictionnaire complet de l'acteur
    return {
        'index': i,
        'name': name,
        'movies': movies_list
    }
```

Write a function `movie_path(origin: int, destination: int) -> distance: int, path: list` that computes the collaboration distance between two actors. That distance is the length of the shortest path `(origin, act1, act2, ..., actX, destination)`, where `origin` and `act` played in the same movie, `act1` and `act2` played in the same movie, ... and
`actX` and `destination` played in the same movie.  In addition to the distance, the response should include one shortest path between the two actors, as a list of the form `["origin_name", "movie1_name", "act1_name", "movie2_name", ..., "destination_name"]`, where `movie1` is a movie that featured `origin` and `act1`, and so on...

In particular:
- One actor is by convention at distance 0 from herself. The return path should be `["origin_name"]` then;
- Two distinct actors that play in the same movie are at distance 1;
- If there is no connection between two actors, the function should return `-1, []` by convention.

**Important remarks**: `movie_path` is tricky. You need to try to implement it but you are allowed to fail. If you are stuck for too long, please explain what you did/try and what blocked you in your opinion. Then move on.

Your answer here.

```python
from collections import deque

def movie_path(origin, destination):
    # --- CAS 1 : L'acteur est lui-même ---
    if origin == destination:
        return 0, [actors[origin][0]]

    # --- INITIALISATION DU BFS ---
    # File d'attente pour explorer les acteurs
    queue = deque([origin])
    
    # Dictionnaire pour mémoriser le chemin : visited[acteur] = (parent, film_qui_lie)
    visited = {origin: (None, None)}
    
    # Ensemble pour ne pas vérifier deux fois le même film (Optimisation cruciale)
    visited_movies = set()

    # --- BOUCLE DE RECHERCHE ---
    while queue:
        current_actor = queue.popleft()

        # On récupère la liste des films de l'acteur courant
        # Structure actors[i] : [Nom, [liste_films]]
        my_movies = actors[current_actor][1]

        for movie_idx in my_movies:
            # Si on n'a jamais exploré ce film
            if movie_idx not in visited_movies:
                visited_movies.add(movie_idx)

                # On récupère tous les acteurs de ce film (le casting)
                # Structure movies[i] : [Titre, [liste_acteurs], Année]
                cast = movies[movie_idx][1]

                for neighbor in cast:
                    if neighbor not in visited:
                        # On note comment on est arrivé là (info parent)
                        visited[neighbor] = (current_actor, movie_idx)

                        # --- VICTOIRE : ON A TROUVÉ LA CIBLE ---
                        if neighbor == destination:
                            # On reconstruit le chemin ici, immédiatement
                            path = []
                            curr = destination
                            
                            # On remonte la chaîne des parents jusqu'à l'origine
                            while curr is not None:
                                # 1. Ajouter le nom de l'acteur
                                path.append(actors[curr][0])
                                
                                # Récupérer les infos du parent
                                parent_idx, link_movie_idx = visited[curr]
                                
                                # 2. Ajouter le nom du film (sauf pour le tout premier nœud qui n'a pas de film parent)
                                if link_movie_idx is not None:
                                    path.append(movies[link_movie_idx][0])
                                
                                # On passe au suivant (en remontant vers l'origine)
                                curr = parent_idx
                            
                            # Le chemin a été construit à l'envers (Arrivée -> Départ), on l'inverse
                            path.reverse()
                            
                            # Calcul de la distance : (Nombre d'éléments - 1) / 2
                            # Ex: [Act1, Film, Act2] = 3 éléments -> (3-1)/2 = 1 saut
                            distance = (len(path) - 1) // 2
                            
                            return distance, path

                        # Si ce n'est pas la cible, on l'ajoute à la file pour explorer ses voisins plus tard
                        queue.append(neighbor)

    # --- ECHEC : AUCUN CHEMIN TROUVÉ ---
    return -1, []
```

```python
from flask import Flask, jsonify, request, abort
from collections import deque

app = Flask(__name__)

def api_error(message, code=400):
    response = jsonify({"error": message})
    response.status_code = code
    return response


def paginate(indices, start, limit):
    return indices[start:start + limit]

@app.route("/movies/<int:id>")
def api_get_movie(id):
    if id < 0 or id >= len(movies):
        return api_error("Movie not found", 404)
    return jsonify(get_movie(id))


@app.route("/movies")
def api_get_movies():
    start = int(request.args.get("start", 0))
    limit = int(request.args.get("limit", 100))
    order = request.args.get("order")

    indices = list(range(len(movies)))

    if order == "name":
        indices.sort(key=lambda i: movies[i][0])
    elif order == "year":
        indices.sort(key=lambda i: movies[i][2])
    elif order is not None:
        return api_error("Invalid order parameter", 400)

    indices = paginate(indices, start, limit)
    return jsonify([get_movie(i) for i in indices])



@app.route("/actors/<int:id>")
def api_get_actor(id):
    if id < 0 or id >= len(actors):
        return api_error("Actor not found", 404)
    return jsonify(get_actor(id))


@app.route("/actors")
def api_get_actors():
    start = int(request.args.get("start", 0))
    limit = int(request.args.get("limit", 100))
    order = request.args.get("order")

    indices = list(range(len(actors)))

    if order == "name":
        indices.sort(key=lambda i: actors[i][0])
    elif order is not None:
        return api_error("Invalid order parameter", 400)

    indices = paginate(indices, start, limit)
    return jsonify([get_actor(i) for i in indices])


@app.route("/actors/<int:id>/costars")
def api_get_costars(id):
    if id < 0 or id >= len(actors):
        return api_error("Actor not found", 404)

    costars = set()

    for movie_index in actors[id][1]:
        for actor_index in movies[movie_index][1]:
            if actor_index != id:
                costars.add(actor_index)

    return jsonify([get_actor(i) for i in costars])


@app.route("/search/actors/<search>")
def api_search_actors(search):
    result = [
        get_actor(i)
        for i in range(len(actors))
        if search.lower() in actors[i][0].lower()
    ]
    return jsonify(result[:100])


@app.route("/search/movies/<search>")
def api_search_movies(search):
    filters = request.args.get("filter")
    indices = []

    for i in range(len(movies)):
        title, _, year = movies[i]
        if search.lower() in title.lower():
            indices.append(i)

    if filters:
        for f in filters.split(","):
            key, value = f.split(":")
            if key == "year":
                indices = [i for i in indices if str(movies[i][2]) == value]

    return jsonify([get_movie(i) for i in indices[:100]])


@app.route("/actors/<int:id_origin>/distance/<int:id_destination>")
def api_actor_distance(id_origin, id_destination):
    if id_origin >= len(actors) or id_destination >= len(actors):
        return api_error("Actor not found", 404)

    graph = {i: set() for i in range(len(actors))}

    for movie in movies:
        actor_list = movie[1]
        for a in actor_list:
            graph[a].update(actor_list)
            graph[a].discard(a)

    queue = deque([(id_origin, [id_origin])])
    visited = set()

    while queue:
        current, path = queue.popleft()
        if current == id_destination:
            return jsonify([len(path) - 1, path])

        visited.add(current)
        for neighbor in graph[current]:
            if neighbor not in visited:
                queue.append((neighbor, path + [neighbor]))

    return api_error("No path found", 404)


if __name__ == "__main__":
    app.run(debug=False, use_reloader=False)
```
